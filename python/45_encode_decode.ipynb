```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
image: np.ndarray = cv2.imread("../images/000000039769.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image)
```

```python
success, buffer = cv2.imencode('.png', image)
success
```

```python
buffer
```

```python
plt.imshow(cv2.imdecode(buffer, cv2.IMREAD_COLOR))      # 彩色
```

```python
plt.imshow(cv2.imdecode(buffer, cv2.IMREAD_GRAYSCALE))  # 灰度
```

```python
plt.imshow(cv2.imdecode(buffer, cv2.IMREAD_UNCHANGED))  # 保持原始格式
```
